```python
# 1. Import Required Libraries
# ==========================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
sns.set_theme(style="whitegrid")
# Display Settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', 1000)

print("="*80)
print("Libraries Imported Successfully")
print("="*80)
```

```python
# 2. Load Dataset from GitHub
# ==========================

# Replace this with your GitHub RAW URL
url = "https://raw.githubusercontent.com/Prajkta11222/Smart-Vehical-Health-Intelligence-and-Predictive-Maintenance-Recommendation-System/refs/heads/main/flex_fuel_predictive_maintenance_dataset.csv"

df = pd.read_csv(url)

print("\nDataset Loaded Successfully!")
print("="*80)

```

```python
# ==========================================================
# STAGE 2 : EXPLORATORY DATA ANALYSIS (EDA)
# PART 1A : Missing Value Analysis
# ==========================================================

print("="*80)
print("EDA PART 1A : MISSING VALUE ANALYSIS")
print("="*80)

# ==========================================================
# Missing Value Count
# ==========================================================

missing_count = df.isnull().sum()
missing_count = missing_count[missing_count > 0].sort_values(ascending=False)

print("\nMissing Value Count")
print("-"*80)

display(missing_count.to_frame(name="Missing Values"))

print("\nObservation:")
print("The above table shows the number of missing values present in each feature.")


# ==========================================================
# Missing Value Percentage
# ==========================================================

missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage = missing_percentage[missing_percentage > 0].sort_values(ascending=False)

print("\nMissing Value Percentage")
print("-"*80)

display(missing_percentage.round(2).to_frame(name="Missing Percentage (%)"))

print("\nObservation:")
print("Most features contain a small percentage of missing values, which can be handled during preprocessing.")


# ==========================================================
# Overall Missing Value Summary
# ==========================================================

total_missing = df.isnull().sum().sum()
total_cells = df.shape[0] * df.shape[1]
overall_missing_percentage = (total_missing / total_cells) * 100

print("\nOverall Missing Value Summary")
print("-"*80)

print(f"Total Missing Values      : {total_missing}")
print(f"Total Dataset Cells       : {total_cells}")
print(f"Overall Missing Percentage: {overall_missing_percentage:.2f}%")

print("\nObservation:")
print("The dataset contains only a small proportion of missing values relative to its total size.")


# ==========================================================
# Missing Value Heatmap
# ==========================================================

plt.figure(figsize=(18, 7))

sns.heatmap(
    df.isnull(),
    cmap="viridis",
    cbar=True,
    yticklabels=False
)

plt.title("Missing Value Heatmap", fontsize=16)
plt.xlabel("Features")
plt.ylabel("Records")

plt.show()

print("Observation:")
print("The heatmap indicates that missing values are sparsely distributed across multiple sensor features rather than concentrated in a single column.")
```

```python
# ==========================================================
# STAGE 2 : EXPLORATORY DATA ANALYSIS (EDA)
# PART 1B : Target Variable Analysis
# ==========================================================

print("="*80)
print("EDA PART 1B : TARGET VARIABLE ANALYSIS")
print("="*80)


# ==========================================================
# Target Variable Count
# ==========================================================

target_count = df["Need_Maintenance"].value_counts()

print("\nTarget Variable Count")
print("-"*80)

display(target_count.to_frame(name="Count"))

print("\nObservation:")
print("The table shows the number of vehicles that require maintenance and those that do not.")


# ==========================================================
# Target Variable Percentage
# ==========================================================

target_percentage = (
    df["Need_Maintenance"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nTarget Variable Percentage")
print("-"*80)

display(target_percentage.to_frame(name="Percentage (%)"))

print("\nObservation:")
print("The percentage distribution indicates whether the dataset is balanced or imbalanced.")


# ==========================================================
# Count Plot
# ==========================================================

plt.figure(figsize=(8,6))

ax = sns.countplot(
    data=df,
    x="Need_Maintenance",
    palette="Set2"
)

plt.title("Distribution of Need_Maintenance", fontsize=16, fontweight="bold")
plt.xlabel("Need Maintenance")
plt.ylabel("Number of Vehicles")

# Add value labels
for container in ax.containers:
    ax.bar_label(container, fmt='%d', fontsize=11)

plt.show()

print("Observation:")
print("The count plot visually represents the distribution of the target classes.")


# ==========================================================
# Pie Chart
# ==========================================================

plt.figure(figsize=(7,7))

plt.pie(
    target_count,
    labels=target_count.index,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.03]*len(target_count),
    shadow=True
)

plt.title("Need_Maintenance Class Distribution", fontsize=16, fontweight="bold")

plt.show()

print("Observation:")
print("The pie chart illustrates the proportion of each target class in the dataset.")


# ==========================================================
# Final Observation
# ==========================================================

print("\n" + "="*80)
print("SUMMARY")
print("="*80)

if target_percentage.max() <= 60:
    print("✓ The target variable is reasonably balanced.")
    print("✓ This is suitable for classification models.")
else:
    print("⚠ The target variable appears imbalanced.")
    print("⚠ Class balancing techniques may be required during preprocessing.")
```

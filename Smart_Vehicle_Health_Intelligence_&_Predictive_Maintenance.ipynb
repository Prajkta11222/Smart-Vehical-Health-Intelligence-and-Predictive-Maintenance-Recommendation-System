#Stage 3 : Data Preprocessing

```python
#Create a Copy of Dataset
print("="*80)
print("STAGE 3 : DATA PREPROCESSING")
print("="*80)

# Keeping the original dataset unchanged
data = df.copy()

print("Dataset copied successfully.")
print("Shape :", data.shape)
```

```python

# Remove Data Leakage Features

leakage_features = [
    "Failure_Type",
    "Remaining_Useful_Life_km"
]

data.drop(columns=leakage_features, inplace=True)

print("Leakage Features Removed")
print(leakage_features)

print("\nCurrent Shape :", data.shape)
```

```python

# Separate Features and Target Variable

X = data.drop("Need_Maintenance", axis=1)

y = data["Need_Maintenance"]

print("Features Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python

# Identify Numerical and Categorical Columns

numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()

print("Numerical Features :", len(numerical_cols))
print(numerical_cols)

print("\nCategorical Features :", len(categorical_cols))
print(categorical_cols)
```

```python
# Handle Missing Values

# Numerical Features
for col in numerical_cols:
    X[col] = X[col].fillna(X[col].median())

# Categorical Features
for col in categorical_cols:
    X[col] = X[col].fillna(X[col].mode()[0])

print("Missing Values After Imputation")
print(X.isnull().sum().sum())
```

```python

# Encode Target Variable
y = y.map({
    "No":0,
    "Yes":1
})

print(y.value_counts())
```

```python
# Label Encoding

from sklearn.preprocessing import LabelEncoder

label_encoders = {}

for col in categorical_cols:

    encoder = LabelEncoder()

    X[col] = encoder.fit_transform(X[col].astype(str))

    label_encoders[col] = encoder

print("Categorical Features Encoded Successfully.")
```

```python

# Train Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("Training Set :", X_train.shape)
print("Testing Set :", X_test.shape)
```
